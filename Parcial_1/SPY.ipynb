```python
import yfinance as yf
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
#bajar datos de yahoo finance
today=pd.Timestamp.today()
start=today - pd.Timedelta(days=365)
start_date=start.strftime('%Y-%m-%d')
start_date
```

```python
symbol="SPY"
df_stock = yf.download(symbol, start=start_date, end=today.strftime('%Y-%m-%d'))[["Close"]].copy()
df_stock
```

```python
#calcular los retornos algoritmicos diarios
#lo vamos a usar para facilitar la simulacion y estadistica
#generar simetria en la distribucion de retornos(NORMAL)
df_stock['Returns']=np.log(df_stock["Close"]/df_stock
["Close"].shift(1))
```
