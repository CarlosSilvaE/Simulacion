```python
import random
```

```python
INTERVAL=10
circle_points=0
square_point=0
```

```python
for i in range(INTERVAL):
    ran_x = random.uniform(-1, 1)
    ran_y = random.uniform(-1, 1)

    #distancia de cada punto en el origen
    origin_distance = ran_x**2 + ran_y**2
    print("Punto: (", ran_x, ", ", ran_y, ")")
    print("Distancia al origen: ", origin_distance)

    if origin_distance <= 1:
        circle_points += 1
    square_point += 1

    pi = 4 * circle_points / square_point
    print("PI estimado: ", pi)
```
