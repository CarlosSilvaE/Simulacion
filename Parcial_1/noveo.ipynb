```python
import random
import matplotlib.pyplot as plt
import numpy as np
```

```python
iteraciones = 10
par1, par2, par3, par4, par5, par6, impar = 0, 0, 0, 0, 0, 0, 0
```

```python
gana_par1 = (1*1) * 100
gana_par2 = (2*2) * 100
gana_par3 = (3*3) * 100
gana_par4 = (4*4) * 100
gana_par5 = (5*5) * 100
gana_par6 = (6*6) * 100
gana_impar =- 200
```

```python
for i in range(iteraciones):
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)

    if dado1 == dado2:
        if dado1 == 1:
            par1 += 1
        elif dado1 == 2:
            par2 += 1
        elif dado1 == 3:
            par3 += 1
        elif dado1 == 4:
            par4 += 1
        elif dado1 == 5:
            par5 += 1
        elif dado1 == 6:
            par6 += 1
    else:
        impar += 1
    print(f"Iteración {i+1}: Dado 1 = {dado1}, Dado 2 = {dado2}")
```

```python
resultado=np.array([impar, par1, par2, par3, par4, par5, par6])
ganancias = np.array([gana_impar, gana_par1, gana_par2, gana_par3, gana_par4, gana_par5, gana_par6])
print(f"Resultados: {resultado}")
print(f"Ganancias: {ganancias}")
```

```python
plt.title("Resultados de los simulacion")
plt.ylabel("Cantidad de tiros")
plt.bar(np.arange(len(resultado)), resultado, color='blue', alpha=0.7, label='Tiros')
cnt=0
for i in resultado:
    plt.text(cnt, i+0.5, str(i), ha='center', va='bottom')
    cnt += 1
plt.show()
```

```python
for i in range(0, 7):
    print(f"Resultado {i}: {resultado[i]} tiros, Ganancia: {ganancias[i]}")
    if i == 0:
        perdidas = resultado[i] * ganancias[i]
    else:
        loquegano = resultado[i] * ganancias[i]
print(f"Total de perdidas: {perdidas}")
print(f"Total de ganancias: {loquegano}")
```
